```python
# One-Sample t-test
from scipy import stats
import matplotlib.pyplot as plt


# Sample data: marks of students
data = [45, 50, 55, 60, 65]

# Test if mean is equal to 50
t_statistic, p_value = stats.ttest_1samp(data, 50)

print("t-test example")
print("t-statistic =", t_statistic)
print("p-value =", p_value)

plt.hist(data, bins=5, color="skyblue", edgecolor="black")
plt.axvline(50, color="red", linestyle="--", label="Population mean=50")
plt.legend()
plt.title("t-test: Sample vs Population Mean")
plt.show()
```
