```python
# One-Sample t-test
from scipy import stats
import matplotlib.pyplot as plt


# Sample data: marks of students
data = [45, 50, 55, 60, 65]

# Test if mean is equal to 50
t_statistic, p_value = stats.ttest_1samp(data, 50)

print("t-test example")
print("t-statistic =", t_statistic)
print("p-value =", p_value)

plt.hist(data, bins=5, color="skyblue", edgecolor="black")
plt.axvline(50, color="red", linestyle="--", label="Population mean=50")
plt.legend()
plt.title("t-test: Sample vs Population Mean")
plt.show()
```

```python
# z-test for proportion
from statsmodels.stats.proportion import proportions_ztest

# Example: 40 students passed out of 100
successes = 40
total = 100
labels = ["Pass", "Fail"]
counts = [40, 60]

# Null hypothesis: proportion = 0.5
z_statistic, p_value = proportions_ztest(successes, total, 0.5)

print("z-test example")
print("z-statistic =", z_statistic)
print("p-value =", p_value)

plt.bar(labels, counts, color=["green", "gray"])
plt.title("z-test: Pass vs Fail Counts")
plt.show()
```

```python
# Chi-Square Test

# Used for categorical data (goodness of fit or independence).

import numpy as np
from scipy.stats import chi2_contingency

# Contingency table
# Gender vs Preference
# Male: 30 like, 10 dislike
# Female: 20 like, 40 dislike
table = np.array([[30, 10],
                  [20, 40]])

chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-Square test example")
print("Chi2 statistic =", chi2)
print("p-value =", p_value)
print("Degrees of freedom =", dof)
print("Expected values =")
print(expected)

# Flatten for plotting
observed = table.flatten()
expected = expected.flatten()
labels = ["Male-Like", "Male-Dislike", "Female-Like", "Female-Dislike"]

# Simple bar chart
plt.bar(labels, observed, label="Observed", color="skyblue")
plt.bar(labels, expected, label="Expected", color="salmon", alpha=0.6)

plt.ylabel("Counts")
plt.title("Chi-Square Test: Observed vs Expected")
plt.legend()
plt.show()
```

```python
# ANOVA (One-way)

# Used when comparing means of more than two groups.

from scipy.stats import f_oneway

# Marks of students in three classes
class_A = [85, 90, 88, 75, 95]
class_B = [70, 65, 80, 72, 68]
class_C = [88, 85, 84, 90, 87]

f_statistic, p_value = f_oneway(class_A, class_B, class_C)

print("ANOVA test example")
print("F-statistic =", f_statistic)
print("p-value =", p_value)

plt.boxplot([class_A, class_B, class_C], labels=["Class A", "Class B", "Class C"])
plt.title("ANOVA: Comparing Class Scores")
plt.show()
```
